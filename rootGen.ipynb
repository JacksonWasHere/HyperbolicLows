```python
load('roots.sage')
system = R3System(2,5,5)
system.gen_smalls();
system.gen_lows();

ordA = int((system.W.coxeter_matrix()[0])[1])
ordB = int(system.W.coxeter_matrix()[0][2])
ordC = int(system.W.coxeter_matrix()[1][2])

import json
outListS = []
for rt in system.Smalls:
    for ft in rt.vector.list():
        outListS.append(float(ft))

outListL = []
for low in system.Lows:
    mLow = system.str_to_word(low)
    for row in matrix(mLow).rows():
        for elt in row.list():
            outListL.append(float(elt))

kvList = [ordA,ordB,ordC]

# PATH TO JSON FILE
# MUST SET NUMBERS MANUALLY
jsPath = "sketch/roots.json"

data = {"kval":kvList,"numSmalls":len(outListS),"numLows":len(outListL),"roots":outListS, "lows": outListL}
with open(jsPath,'w') as f:
    json.dump(data, f, indent = 4)
```
